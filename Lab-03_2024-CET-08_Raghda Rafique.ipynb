```python
import pandas as pd

cars = pd.read_csv("pakistan_used_cars.csv")

print("First 5 Records:")
print(cars.head())

print("Dataset Size:")
print(cars.shape)

print("Column Names:")
print(cars.columns.tolist())

print("Dataset Information:")
cars.info()
```

```python
# Task 2

cars.columns = cars.columns.str.strip()

# A) Select one column
print("Brand:")
print(cars["Brand"].head())

# B) Select rows
print("\nFirst 10 Rows:")
print(cars.iloc[:10])

# C) Select specific columns
print("\nSelected Columns:")
print(cars[["Brand", "Model", "Price"]].head())

# D) Filter Toyota cars
print("\nToyota Cars:")
print(cars[cars["Brand"] == "Toyota"].head())

# E) Filter cars above 10 lakh
print("\nCars Above 10 Lakh:")
price = pd.to_numeric(
    cars["Price"].astype(str).str.replace(",", "").str.replace("Rs", ""),
    errors="coerce"
)
print(cars[price > 1000000].head())

# F) Sort by price
print("\nHighest Price Cars:")
print(cars.loc[price.sort_values(ascending=False).index].head())
```

```python
# Task 3: Data Cleaning

import pandas as pd
import numpy as np

# A) Remove missing Brand and Model
cars = cars.dropna(subset=["Brand", "Model"])

# Fill missing text values
for col in cars.select_dtypes(include="object").columns:
    cars[col] = cars[col].fillna("Unknown")

# B) Clean city names
cars["Registered City"] = cars["Registered City"].astype(str).str.strip().str.title()

# C) Clean Price
cars["Price"] = cars["Price"].astype(str).str.replace(
    r"[^0-9]", "", regex=True
)
cars["Price"] = pd.to_numeric(cars["Price"], errors="coerce")

# D) Convert Year
cars["Year"] = pd.to_numeric(cars["Year"], errors="coerce").astype("Int64")

# E) Correct spelling
cars["Registered City"] = cars["Registered City"].replace("Sukkar", "Sukkur")
cars["Model"] = cars["Model"].str.replace("Corrolla", "Corolla", regex=False)

# F) Remove duplicate rows
cars = cars.drop_duplicates()

# G) Clean KMs Driven
cars["KMs Driven"] = pd.to_numeric(
    cars["KMs Driven"].astype(str).str.replace(",", ""),
    errors="coerce"
)
cars.loc[cars["KMs Driven"] < 1000, "KMs Driven"] = np.nan

# H) Remove invalid prices
cars = cars.dropna(subset=["Price"])
cars = cars[cars["Price"] >= 50000]
cars["Price"] = cars["Price"].astype("int64")

# Final Result
print("Final Dataset Shape:", cars.shape)
print(cars.isnull().sum())
print(cars.dtypes)
```

```python
# Task 4: Data Analysis

# A) Price in lakh
cars["Price_lakh"] = cars["Price"] / 100000
print("A) Price in Lakh:")
print(cars["Price_lakh"].head())

# B) Car age
cars["Age"] = 2020 - cars["Year"]
print("\nB) Car Age:")
print(cars["Age"].head())

# C) Top 5 brands
print("\nC) Top 5 Brands:")
print(cars["Brand"].value_counts().head(5))

# D) Top 5 cities
print("\nD) Top 5 Cities:")
print(cars["Registered City"].value_counts().head(5))

# E) Median price by fuel
print("\nE) Median Price by Fuel:")
print(cars.groupby("Fuel")["Price"].median())

# F) Corolla GLI prices from 2010 to 2018
print("\nF) Corolla GLI Prices:")
corolla = cars[
    cars["Model"].str.contains("Corolla GLI", case=False, na=False)
]
corolla = corolla[corolla["Year"].between(2010, 2018)]
year_price = corolla.groupby("Year")["Price"].median()
print(year_price)

print("Yearly Price Change:")
print(year_price.diff())

# G) Five most expensive cars
print("\nG) Five Most Expensive Cars:")
print(cars.nlargest(5, "Price")[["Brand", "Model", "Year", "Price"]])

# H) Cash vs Installment
print("\nH) Cash vs Installment:")
print(cars.groupby("Transaction Type")["Price"].agg(["count", "median"]))
```

```python
# Task 5: Save Clean Dataset

cars.to_csv("pakistan_used_cars_clean.csv", index=False)

print("Clean dataset saved successfully!")
print("Dataset Shape:", cars.shape)
```
