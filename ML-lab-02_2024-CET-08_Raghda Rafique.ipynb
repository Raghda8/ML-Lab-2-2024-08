```python
import numpy as np

attendance = np.zeros(10)
fee_paid = np.ones(10)
grace_marks = np.full(10, 5)

roll_numbers = np.arange(10, 51, 10)
even_numbers = np.arange(2, 21, 2)

seat_map = np.arange(9).reshape(3, 3)
identity = np.eye(3)

print("Attendance:", attendance)
print("Fee Paid:", fee_paid)
print("Grace Marks:", grace_marks)
print("Roll Numbers:", roll_numbers)
print("Even Numbers:", even_numbers)
print("Seat Map:\n", seat_map)
print("Identity Matrix:\n", identity)
```

```python
import numpy as np

a = np.linspace(0.01, 1.00, 100)
b = np.linspace(1, 20, 20)

rain = np.array([10, 20, 30, 50, 100, 150,
                 200, 180, 74, 40, 15, 5])

quarter = rain.reshape(4, 3)

print("10x10 Array:")
print(a.reshape(10, 10))

print("20 Points:")
print(b)

print("Quarterly Rainfall:")
print(quarter)

total = quarter.sum(axis=1)
print("Quarterly Total:", total)

print("Wettest Quarter:", total.argmax() + 1)
```

```python
import numpy as np

hall = np.arange(1, 26).reshape(5, 5)

print("Seating Plan:")
print(hall)

print("Selected Seats:")
print(hall[2:, 1:])

print("Single Seat:")
print(hall[3, 4])

print("First Three Rows:")
print(hall[0:3, 1:2])

print("Last Row:")
print(hall[4])

print("Last Two Rows:")
print(hall[3:])

print("Total Seats:", hall.sum())
print("Standard Deviation:", hall.std())
```

```python
import numpy as np

cities = np.array(['Lahore', 'Karachi', 'Multan', 'Peshawar'])
population = np.array([130, 160, 22, 20])

print("Cities:", cities)
print("Population:", population)
print("Total Population:", np.sum(population))
print("Highest Population:", np.max(population))
print("Lowest Population:", np.min(population))
print("Average Population:", np.mean(population))
```

```python
import numpy as np

prices = np.array([50, 60, 70, 80, 90])

new_prices = prices + 15

print("Old Prices:", prices)
print("New Prices:", new_prices)
print("Total:", np.sum(new_prices))
print("Average:", np.mean(new_prices))
print("Highest Price:", np.max(new_prices))
```

```python
import numpy as np

marks = np.array([28, 32, 25, 35, 30])

grace_marks = marks + 5

print("Original Marks:", marks)
print("Grace Marks:", grace_marks)

passed = grace_marks >= 30

print("Passed Students:", np.sum(passed))
print("Failed Students:", np.sum(~passed))
```
